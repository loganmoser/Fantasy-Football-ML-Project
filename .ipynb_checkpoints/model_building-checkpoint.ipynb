```python
# Import packages
import nflreadpy
import polars as pl
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
player_data = nflreadpy.load_player_stats(seasons=True, summary_level="reg")

# Select unnecessary columns
drop_columns = ["headshot_url", "season_type", "gwfg_made", "gwfg_att", "gwfg_missed",
                    "gwfg_blocked", "def_tackles_solo", "def_tackles_with_assist",
                    "def_tackle_assists", "def_tackles_for_loss", "def_tackles_for_loss_yards",
                    "def_fumbles_forced", "def_sacks", "def_sack_yards", "def_qb_hits",
                    "def_interceptions", "def_interception_yards", "def_pass_defended",
                    "def_tds", "def_fumbles", "def_safeties", "punt_returns", "fumble_recovery_own",
                    "fumble_recovery_yards_own", "fumble_recovery_opp", "fumble_recovery_yards_opp",
                    "fumble_recovery_tds", "punt_return_yards", "kickoff_returns", "kickoff_return_yards"]

player_data_cleaned = player_data.drop(drop_columns)
# Compute each player's first season in the dataset (their rookie year)
rookie_seasons = (
    player_data_cleaned
    .group_by("player_display_name")
    .agg(pl.col("season").min().alias("rookie_season"))
)

# Join this back onto the main data
player_data_cleaned = player_data_cleaned.join(
    rookie_seasons, on="player_display_name", how="left"
)

# Years of experience = current season - rookie season
player_data_cleaned = player_data_cleaned.with_columns(
    (pl.col("season") - pl.col("rookie_season")).alias("years_experience")
)
# Positional columns
qb_cols = ["player_display_name", "season", "completions", "attempts", "passing_yards", "passing_tds",
               "passing_interceptions", "sacks_suffered", "sack_yards_lost", "sack_fumbles",
               "sack_fumbles_lost", "passing_air_yards", "passing_yards_after_catch",
               "passing_first_downs", "passing_epa", "passing_cpoe", "passing_2pt_conversions",
               "pacr", "carries", "rushing_yards", "rushing_tds", "rushing_fumbles",
               "rushing_fumbles_lost", "rushing_first_downs", "rushing_epa",
               "rushing_2pt_conversions", "years_experience", "fantasy_points_ppr"]
# RBs, TEs, and WRs had many similar columns so they are combined into skill-position columns
sp_cols = ["player_display_name", "carries", "rushing_yards", "rushing_tds", "rushing_fumbles", 
               "rushing_fumbles_lost", "rushing_first_downs", "rushing_epa", "rushing_2pt_conversions",
               "receptions", "targets", "receiving_yards", "receiving_tds", "receiving_fumbles",
               "receiving_fumbles_lost", "receiving_air_yards", "receiving_yards_after_catch",
               "receiving_first_downs", "receiving_epa", "receiving_2pt_conversions", "racr",
               "target_share", "air_yards_share", "wopr", "years_experience", "fantasy_points_ppr"]
   

# Group players by position to build positional models
qbs = player_data_cleaned.filter(pl.col('position') == 'QB')[qb_cols]
rbs = player_data_cleaned.filter(pl.col('position') == 'RB')[sp_cols]
tes = player_data_cleaned.filter(pl.col('position') == 'TE')[sp_cols]
wrs = player_data_cleaned.filter(pl.col('position') == 'WR')[sp_cols]
```

```python
"""
Feature Engineering for QBs
- Completion %
- yds/att
- td/int ratio
- sacks/game
- VORP (Value of Replacement Player)

Features for RBs

Features for TEs

Features for WRs
"""
qbs_after_features_drop = ["completions", "attempts", "sack_yards_lost", "sack_fumbles", "passing_air_yards", "passing_yards_after_catch", "passing_first_downs", "rushing_fumbles", "rushing_first_downs", "rushing_2pt_conversions"]


qbs = qbs.with_columns(
    (pl.col("completions") / pl.col("attempts")).alias("completion_percentage").replace([float("inf"), float("-inf")], 0),
    (pl.col("passing_yards") / pl.col("attempts")).alias("yards_per_attempt").replace([float("inf"), float("-inf")], 0),
    (pl.col("attempts") / pl.col("passing_tds")).alias("TDS/Pass").replace([float("inf"), float("-inf")], 0),
    (pl.col("passing_tds") / pl.col("passing_interceptions")).alias("interception_ratio").replace([float("inf"), float("-inf")], 0),
    (pl.col("sacks_suffered") / 18).alias("sacks_per_game").replace([float("inf"), float("-inf")], 0),
)

#qbs = qbs.drop(qbs_after_features_drop)
```

```python
"""
This will be an ensemble model analysis that will be done using a variety of models: random forest, clustering, and linear regression
will be trained on the data and the predictions of these models will be fed along with original data into a neural network
"""
# Training linear regression model
from sklearn.linear_model import SGDRegressor, Ridge
from sklearn.model_selection import RandomizedSearchCV, train_test_split, cross_val_predict
from sklearn.metrics import auc, accuracy_score, root_mean_squared_error, mean_squared_error, silhouette_score
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import StandardScaler, normalize
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.base import clone
```

```python
# Data Preprocessing Pipeline
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")), 
    ("standardizer", StandardScaler()),
])
```

```python
def build_prediction_dataset(
    df: pl.DataFrame,
    target_col: str = "fantasy_points_ppr",
    id_cols: list[str] = ["player_display_name", "season"],
    lag: int = 1
):
    """
    Pairs each season's stats with the NEXT season's target.
    
    2024 stats → 2025 fantasy points
    2023 stats → 2024 fantasy points
    ...all the way back to 1999 stats → 2000 fantasy points
    """
    
    # Everything that isn't an ID or the target is a feature
    feature_cols = [c for c in df.columns if c not in id_cols + [target_col]]
    
    # Left side: the stats (features) — unchanged
    stats = df.select(*id_cols, *feature_cols)
    
    # Right side: the target — shift season back so it aligns with prior year
    targets = df.select(
        "player_display_name",
        (pl.col("season") - lag).alias("season"),  # 2025 becomes 2024, etc.
        target_col
    )
    
    # Join: every year's stats get paired with next year's outcome
    combined = stats.join(
        targets,
        on=["player_display_name", "season"],
        how="inner"  # drops players who don't appear in consecutive years
    )
    
    return combined


# --- Usage ---
combined = build_prediction_dataset(qbs)

# See what you got — one row per player per year pair
print(combined.shape)
print(combined.select("player_display_name", "season", "fantasy_points_ppr").sample(10))

X, y = combined.drop(pl.col(["player_display_name","fantasy_points_ppr"])), combined["fantasy_points_ppr"]

X_cols = X.columns
```

```python
train_mask = X["season"] <= 2016
validation_mask = (X["season"] <=2020) & ~(train_mask)
test_mask = ~(train_mask) & ~(validation_mask)

X_train, y_train = X.filter(train_mask), y.filter(train_mask)
X_valid, y_valid = X.filter(validation_mask), y.filter(validation_mask)
X_test, y_test = X.filter(test_mask), y.filter(test_mask)

X_train = pl.DataFrame(num_pipeline.fit_transform(X_train), schema=X_cols)
X_valid = pl.DataFrame(num_pipeline.transform(X_valid), schema=X_cols)
X_test = pl.DataFrame(num_pipeline.transform(X_test), schema=X_cols)
```

```python
# Find best number of clusters for kmeans
scores = []
num_clusters = np.arange(2,11)
models = []
X_copy = X_train
pca = PCA(n_components=0.75)
X_copy = pca.fit_transform(X_copy)

for n in num_clusters:
    kmeans = KMeans(n_clusters=n, n_init=10, random_state=42)
    kmeans.fit(X_copy)
    models.append(kmeans)
    scores.append(silhouette_score(X_copy, kmeans.labels_))

scores_df = pl.DataFrame({ "clusters": num_clusters, "scores": scores, "models": models})

best_model = scores_df.sort("scores", descending=True)["models"][0] # This is our best model that will give us the clusters

train_distances = best_model.transform(X_copy) # Final model predictions that will be used as a feature

sns.lineplot(data=scores_df[["clusters", "scores"]], x="clusters", y="scores")
plt.title("Shilouette Scores")
plt.show()
```

```python
X_train_clusters = pl.DataFrame(best_model.transform(X_copy))
X_train_clusters.columns = [f"cluster_{x+1}_distance" for x in range(best_model.n_clusters)]
X_train = X_train.hstack(X_train_clusters)

X_valid_clusters = pl.DataFrame(
    best_model.transform(
        pca.transform(
           X_valid
        )
    )
)
X_valid_clusters.columns = [f"cluster_{x+1}_distance" for x in range(best_model.n_clusters)]
X_valid = X_valid.hstack(X_valid_clusters)

X_test_clusters = pl.DataFrame(
    best_model.transform(
        pca.transform(
            X_test
        )
    )
)
X_test_clusters.columns = [f"cluster_{x+1}_distance" for x in range(best_model.n_clusters)]
X_test = X_test.hstack(X_test_clusters)
```

```python
def base_model_predictions(X_train, X_test, X_valid, y_train, model_name, model, params = {}):
    pipe = Pipeline([(model_name, model)])
    search = RandomizedSearchCV(pipe, param_distributions=params,
                                cv=10, n_jobs=-1, random_state=42)
    search.fit(X_train, y_train)
    best = search.best_estimator_

    # meta-feature for TRAIN: out-of-fold preds (no leakage)
    train_preds = pl.DataFrame(cross_val_predict(clone(best), X_train, y_train, cv=10, n_jobs=-1))
    train_preds.columns = [f"{str(model).split('(')[0]}_predictions"] # to identify which model made this prediction

    #meta-features for VALID:
    valid_preds = pl.DataFrame(cross_val_predict(clone(best), X_valid, y_valid, cv=10, n_jobs=-1))
    valid_preds.columns = [f"{str(model).split('(')[0]}_predictions"] # to identify which model made this prediction

    # meta-feature for TEST: model trained on full train, applied to unseen test
    test_preds = pl.DataFrame(best.predict(X_test))
    test_preds.columns = [f"{str(model).split('(')[0]}_predictions"] # to identify which model made this prediction

    return train_preds, test_preds, valid_preds, best
```

```python
sgd_params = {"sgd__loss": ["squared_error", "huber", "epsilon_insensitive", "squared_epsilon_insensitive"],
              "sgd__penalty": ["l2", "l1", "elasticnet", None], "sgd__alpha": np.arange(1e-6, 0.1, 0.001),
              "sgd__learning_rate": ["constant", "optimal", "invscaling", "adaptive", "pa1", "pa2"],
              "sgd__early_stopping": [True, False], "sgd__warm_start": [True, False]}
rf_params = {"rf__loss": ["squared_error", "absolute_error", "huber", "quantile"], "rf__learning_rate": np.arange(1e-3, 3, 0.01),
             "rf__n_estimators": np.arange(100, 1000, 200), "rf__criterion": ["friedman_mse", "squared_error"],
             "rf__max_depth": np.arange(3, 20, 1), "rf__max_features": [None, "sqrt", "log2"], "rf__warm_start": [True, False]}

lr_train, lr_test, lr_valid, lr = base_model_predictions(X_train, X_test, X_valid, y_train, "lr", LinearRegression())
X_train, X_test, X_valid = X_train.hstack(lr_train), X_test.hstack(lr_test), X_valid.hstack(lr_valid)
sgd_train, sgd_test, sgd_valid, sgd = base_model_predictions(X_train, X_test, X_valid, y_train, "sgd", SGDRegressor(max_iter=20_000), sgd_params)
X_train, X_test, X_valid = X_train.hstack(sgd_train), X_test.hstack(sgd_test), X_valid.hstack(sgd_valid)
rf_train, rf_test, rf_valid, rf = base_model_predictions(X_train, X_test, X_valid, y_train, "rf", GradientBoostingRegressor(random_state=42), rf_params)
X_train, X_test, X_valid = X_train.hstack(rf_train), X_test.hstack(rf_test), X_valid.hstack(rf_valid)
```

```python
import tensorflow as tf
```

```python
# keras can't work with polars or pandas dataframes and needs to be a numpy array
X_train_np = X_train.to_numpy()
X_valid_np = X_valid.to_numpy()
y_train_np = y_train.to_numpy()
y_valid_np = y_valid.to_numpy()
X_test_np = X_test.to_numpy()
y_test_np = y_test.to_numpy()
```

```python
# Build Neural network
tf.keras.backend.clear_session()
tf.random.set_seed(42)
normalizer = tf.keras.layers.Normalization(input_shape=X_train_np.shape[1:])
normalizer.adapt(X_train_np)

model = tf.keras.Sequential([
    #normalizer,
    tf.keras.layers.Dense(64, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(0.01)),
    tf.keras.layers.Dense(32, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(0.01)),
    tf.keras.layers.Dense(16, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(0.01)),
    tf.keras.layers.Dropout(0.1),
    tf.keras.layers.Dense(1) # Output
])
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)
model.compile(loss="mse", optimizer=optimizer, metrics=["RootMeanSquaredError"])
early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
history = model.fit(X_train_np, y_train_np, epochs=50, validation_data=(X_valid_np, y_valid_np), callbacks=[early_stop])
model.summary()
```

```python
val_predictions = model.predict(X_valid_np[:10])
val_true = y_valid_np[:10]
```

```python
plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.legend()
plt.show()
```

```python
plt.scatter(range(len(val_predictions)), val_predictions, color='blue')
plt.scatter(range(len(val_true)), val_true, color='red')
plt.legend(['predictions', 'true values'])
```

```python
# Build wide-deep NN
X_train_wide = X_train_np[:, :15]
X_train_deep = X_train_np[:, 10:]
X_valid_wide = X_valid_np[:, :15]
X_valid_deep = X_valid_np[:, 10:]
X_test_wide = X_test_np[:, :15]
X_test_deep = X_test_np[:, 10:]

tf.keras.backend.clear_session()
tf.random.set_seed(42)

input_wide = tf.keras.layers.Input(shape=[15])
input_deep = tf.keras.layers.Input(shape=[26])
norm_layer_wide = tf.keras.layers.Normalization()
norm_layer_deep = tf.keras.layers.Normalization()
norm_wide = norm_layer_wide(input_wide)
norm_deep = norm_layer_deep(input_deep)
hidden1 = tf.keras.layers.Dense(64, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(0.01))(norm_deep)
hidden2 = tf.keras.layers.Dense(32, activation="relu", kernel_regularizer=tf.keras.regularizers.l2(0.01))(hidden1)
concat = tf.keras.layers.concatenate([norm_wide, hidden2])
output = tf.keras.layers.Dense(1)(concat)
wd_model = tf.keras.Model(inputs=[input_wide, input_deep], outputs=[output])

wd_optimizer = tf.keras.optimizers.SGD(learning_rate=1e-4, momentum=0.15)

wd_model.compile(loss="mse", optimizer=wd_optimizer, metrics=["RootMeanSquaredError"])

norm_layer_wide.adapt(X_train_wide)
norm_layer_deep.adapt(X_train_deep)
wd_history = wd_model.fit((X_train_wide, X_train_deep), y_train, epochs=200,
                          validation_data=((X_valid_wide, X_valid_deep), y_valid), callbacks=[early_stop])
```

```python
wd_predictions = wd_model.predict((X_test_wide[:10], X_test_deep[:10]))
wd_true = y_test[:10]

plt.scatter(range(len(wd_predictions)), wd_predictions, color='blue')
plt.scatter(range(len(wd_true)), wd_true, color='red')
plt.legend(['predictions', 'true values'])
```
